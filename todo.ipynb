```python
from datetime import date
task={}
print("*****************To DO List*****************")
print("********************************************")
while True:
    ch=int(input("Enter 1.To Add your Task \n2.To view your tasks \n3.To Delete your tasks \n4.To update your Task"))
    if(ch==1):
        task_title=input("Enter task title")
        tasks=input("Enter task")
        ddate=date.today()

        task[task_title]={   
        "task":tasks,
        "task date":ddate,   
      }
        task_completion=input("Enter if the task is completed(Y/N)")
        is_complete=task[task_title].setdefault("completed","completed" if task_completion in["", "y"]else "Not Completed")
        
    if(ch==2):
        for title,task_det in task.items():
            print(title,end=" ")
            for value in task_det.values():
                 print(value,sep=" ",end=" ")
            print()
    if(ch==3):
        delete_it=input("Enter the task title to remove:")
        res=task.pop(delete_it,False)
        if(res):
            print("Successfully")
        else:
            print("Not available")
    
    if(ch==4):
        taskk_title=input("Enter the task title to update")
        key=input("Enter the key you want to update")
        value=input("Enter the new value ")
        if(taskk_title in task): 
            if(key in task[taskk_title]):
                task[taskk_title].update({key:value})
                print("Task updated successfully")
            else:
                print("key does not exist")
        else:
            print("Task not found")
    op=input("Continue(Y/N)")
    if(op=="N"):
        break
```
